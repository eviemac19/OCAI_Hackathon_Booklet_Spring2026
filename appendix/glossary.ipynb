# Clinical AI Glossary 

The glossary is built around a single simulated cardiovascular risk dataset so that the terms do not feel abstract. The organising principle is deliberately physiological: AI systems are taught as if they were clinical systems with sensing, representation, learning, inference, evaluation and governance.

The same dataset is used repeatedly to explain features, model training, inference, performance metrics, bias, calibration and governance.

The ICO framing is used as the broad working definition because it is practical for NHS clinical governance: AI is treated as an umbrella term for algorithm-based technologies that solve complex tasks by carrying out functions that previously required human thinking, including systems that support predictions, classifications, recommendations and decisions with or without a human in the loop.

## Illustrative Simulated Dataset

The dataset is not intended to represent a real patient dataset; it is a teaching device that makes AI terminology concrete.

**Outcome variable:** 5-year cardiac mortality.

**Candidate predictors include:**

- Demographics: age, sex, ethnicity, deprivation index.
- Risk factors: smoking, diabetes, hypertension, chronic kidney disease, previous myocardial infarction.
- Observations: systolic blood pressure, BMI.
- Bloods: LDL, HDL, HbA1c, eGFR, troponin.
- ECG features: rhythm, QRS duration, ST-T abnormality.
- Imaging: echocardiographic left ventricular function.
- Medications: statin, antiplatelet, beta blocker, ACE inhibitor or ARB.

A classical risk-score approach, such as a QRISK-style model, uses a limited number of prespecified variables with relatively transparent weighting. An AI model can ingest a broader feature set, including nonlinear interactions, longitudinal trends, waveforms and unstructured text. 

That power may improve prediction, but it also increases the need for calibration, external validation, explainability and governance.

## Worked example

**Patient:** 68-year-old man with current smoking, type 2 diabetes, hypertension, systolic blood pressure 158 mmHg, LDL 4.2 mmol/L, mildly elevated troponin, nonspecific ST-T changes on ECG and mild left ventricular impairment on echocardiography.

### Classical QRISK-style model

A QRISK-style model uses a limited set of established cardiovascular risk factors, with transparent coefficients and a clinically familiar output. It is valuable precisely because it is simple, auditable and interpretable. It's also deterministic, with a clear formula which guarantees that inputs are linked to outputs in a very explainable way.  https://qrisk.org/ 

For this patient, the classical model might generate an estimated cardiovascular risk around **12% over 10 years**. This is useful for population-level preventive care and supports a NICE-style threshold-based decision about lipid lowering and risk factor modification.

### AI model approach

An AI model can incorporate a wider set of information: troponin, echo findings, longitudinal blood pressure, renal function, medication history, deprivation index, ECG features and interactions between these variables. It may also incorporate wearables data e.g. the patients' fitbit readings from the past month. You might even want to build in dynamic data from the patients' exercise tolerance test. Using multiple, structured and unstructured, complex inputs,  AI can be used to model nonlinear risk, such as the way diabetes, age, renal function and LV impairment may combine to produce higher-than-additive risk.

For the same patient, the AI model might estimate **22% 5-year cardiac mortality risk**.

AI models are typically probabilistic. This means the output is a probably one, rather than a easily calculated, deterministic one. We can set constraints to ensure that the most probably output is presented,however, this does not guarantee that this is the case. This makes AI explainability quite a complex and interesting field, and part of clinical risk management. 

### Teaching point

The AI model is not simply a faster QRISK. It is a different kind of system. The classical score is deliberately constrained, deterministic and interpretable; the AI model is broader, more expressive and potentially more accurate, but also more dependent on validation, calibration, subgroup performance, monitoring and governance.

## Performance: evaluation

<details open>
<summary>Expand section</summary>

| Term | Explanation | Illustration |
|---|---|---|
| Accuracy | The proportion of predictions that are correct. Accuracy is intuitive but often misleading in imbalanced clinical datasets, where a model can appear accurate by predicting the majority class while missing important minority outcomes. | <details><summary>Open</summary>If cardiac death is uncommon, a model that predicts survival for almost everyone may have high accuracy but poor clinical value.</details> |
| Sensitivity | The proportion of true positive cases correctly identified by the model. High sensitivity is important when missing cases carries significant clinical risk. | <details><summary>Open</summary>The proportion of patients who die from cardiac causes who were correctly identified as high risk.</details> |
| Specificity | The proportion of true negative cases correctly identified. High specificity reduces unnecessary interventions, anxiety and workload. | <details><summary>Open</summary>The proportion of patients who do not die who are correctly classified as low risk.</details> |
| Precision | The proportion of positive predictions that are correct. This reflects the reliability of identifying high-risk patients and is strongly influenced by outcome prevalence. | <details><summary>Open</summary>Among those predicted to be high risk, the percentage who actually experience cardiac mortality.</details> |
| Recall | Another term for sensitivity. It is commonly used in machine learning literature and refers to the ability to retrieve or identify true cases. | <details><summary>Open</summary>In the cardiac project, recall measures how many future cardiac deaths the model successfully flags.</details> |
| F1 score | A single metric combining precision and recall. It is useful when classes are imbalanced, but it does not replace clinical judgement about the relative cost of false positives and false negatives. | <details><summary>Open</summary>A model with balanced precision and sensitivity may be preferable when both missed deaths and unnecessary escalation are clinically important.</details> |
| ROC curve | A graphical representation of the trade-off between sensitivity and specificity across different thresholds. | AUC | A summary measure of a model's ability to discriminate between those who experience the outcome and those who do not across all thresholds. # QRISK-style model versus AI model: the core comparison

| Dimension | QRISK-style risk score | AI mortality model |
|---|---|---|
| Data inputs | Limited, prespecified clinical variables | Broader multimodal data, including bloods, ECG, echo, temporal trends and text-derived features |
| Relationship form | Mostly fixed and transparent | Can learn nonlinear and high-dimensional interactions |
| Interpretability | High | Variable; may require explanation tools |
| Clinical familiarity | Strong | Depends on training, implementation and education |
| Performance potential | Reliable baseline, constrained feature space | Potentially higher discrimination and personalisation |
| Main risk | Oversimplification and limited individualisation | Opaqueness, bias, drift, leakage and over-reliance |
| Governance need | Established clinical audit and updating | Stronger assurance, external validation, calibration, subgroup analysis and monitoring |

The teaching point is not that AI is automatically superior. The teaching point is that AI changes the balance of strengths and risks. It may detect clinically meaningful patterns beyond traditional scores, but its use must be justified by performance, calibration, fairness, safety and clinical workflow benefit.

# Exam Pearls

<details open>
<summary>Expand section</summary>

| Pitfall | Explanation | Cardiac project illustration |
|---|---|---|
| Confusing accuracy with clinical usefulness | Accuracy can be misleading when outcomes are uncommon. A model may be numerically accurate while failing the patients who matter most. | If most patients survive, predicting survival for everyone gives high accuracy but misses cardiac deaths. |
| Ignoring calibration | A model can discriminate well but produce risk probabilities that are too high or too low. Calibration matters when thresholds drive treatment. | A 20% predicted risk should correspond to about 20% observed mortality among similar patients. |
| Treating AUC as the whole answer | AUC measures discrimination, not safety, calibration, fairness or clinical utility. | An AUC of 0.85 is promising but not sufficient for deployment. |
| Missing data leakage | Leakage makes models appear excellent in development and unsafe in practice. | Post-event troponin values must not be used to predict pre-event mortality risk. |
| Assuming explainability equals correctness | An explanation can make a wrong prediction feel plausible. Explainability supports scrutiny but does not prove validity. | A model may cite troponin and LV impairment in a persuasive explanation while still being poorly calibrated. |
| Assuming AI replaces judgement | AI outputs require interpretation, context and accountability. | The clinician must decide management after considering symptoms, comorbidity, preferences and guidance. |
| Ignoring subgroup performance | Overall performance can mask harm in specific groups. | The model may perform well overall but underestimate risk in women, minority ethnic groups or deprived populations. |
| Forgetting drift | A model that was valid at deployment may become invalid as practice changes. | New lipid-lowering therapies or care pathways may reduce observed mortality over time. |
| Treating thresholds as technical facts | Thresholds represent clinical, ethical and policy choices. | A 10% intervention threshold reflects risk tolerance, capacity, treatment burden and benefit. |
